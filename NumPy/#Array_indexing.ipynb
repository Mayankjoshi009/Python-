Numpy Indexing And Slicing Of Vectors

```python
import numpy as np
```

```python
arr =np.arange(11,21)
arr
```

```python
arr[9]
```

```python
arr[2:9]
```

```python
arr[2:9:2]
```

```python
arr_slice=arr[3::2]
arr_slice
```

```python
arr[9]
```

Numpy Indexing And Slicing Of Matrix

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[5]
```

```python
arr[5,4]
```

```python
arr[0,2]
```

```python
slice=arr[2:5,3:5]
slice
```

```python
slice1=arr[0:3,0:3]
slice1
```

```python
slice4=arr[3:6,0:3]
slice4
```

```python
arr[:,2]
```

Boolean Indexing

```python
arr=np.arange(11,21)
arr
```

```python
bool_index = arr%2==0
bool_index
```

```python
arr=arr[bool_index]
arr
```
