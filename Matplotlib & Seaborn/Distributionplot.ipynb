```python
df=sns.load_dataset('tips')
```

```python


plt.subplot(1, 2, 1)
sns.histplot(data=df, x='total_bill', kde=True, bins=20)

plt.subplot(1, 2, 2)
sns.histplot(data=df, x='tip', kde=True, bins=20)

```
