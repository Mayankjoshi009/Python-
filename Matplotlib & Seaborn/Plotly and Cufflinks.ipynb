```python
import pandas as pd
import numpy as np
import seaborn as sns
import cufflinks as cf 
cf.go_offline()
from plotly.offline import iplot
import plotly.express as px
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

```python


px.scatter(tips, x='total_bill', y='tip').show()
```

```python
px.line(tips, y="total_bill", title="Total Bill").update_traces(line_color="purple").show()
```

```python
day_tip_mean = tips.groupby('day')['tip'].mean().reset_index()

fig = px.bar(
    day_tip_mean,
    x='day',
    y='tip',
    title='Average Tip by Day',
    color_discrete_sequence=["#E421DA"]
    
)
fig.update_traces(marker_line_color='rgba(0,0,0,0.4)', marker_line_width=1)
fig.show()
```

```python
fig = px.scatter(
    tips,
    x='total_bill',
    y='tip',
    color='sex',
    color_discrete_sequence=['#E421DA', '#2E86DE'],
    title='Total Bill vs Tip'
)

fig.update_traces(marker=dict(size=10))
fig.show()
```

```python
tips.groupby('day')['tip'].mean().iplot(kind = 'bar')
```
