```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Matplotlib Basic

```python
x = np.linspace(0,5,11)
x 
y=x**2
y
```

```python
plt.plot(x,y)
plt.title("Graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(y**6,x**3)
plt.subplot(2,2,2)
plt.plot(x**2,x**2)
```

```python
plt.plot(x, y)
plt.title("y = x²")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
